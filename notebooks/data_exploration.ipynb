# Book Genre Classifier - Data Exploration

This notebook explores the book description dataset and analyzes genre distributions, text characteristics, and potential challenges.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
import sys
sys.path.append('../')

from config import *

plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

## Load and Basic Analysis

```python
# Load data
df = pd.read_csv(RAW_DATA_FILE)
print(f"Dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df.head()
```

```python
# Genre distribution
plt.figure(figsize=(12, 6))
genre_counts = df['genre'].value_counts()
sns.barplot(x=genre_counts.values, y=genre_counts.index)
plt.title('Distribution of Book Genres')
plt.xlabel('Number of Books')
plt.tight_layout()
plt.show()

print(f"Number of genres: {df['genre'].nunique()}")
print(f"Genre distribution:\n{genre_counts}")
```

## Text Analysis

```python
# Description length analysis
df['description_length'] = df['description'].str.len()
df['word_count'] = df['description'].str.split().str.len()

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Character length distribution
axes[0,0].hist(df['description_length'], bins=50, alpha=0.7)
axes[0,0].set_title('Distribution of Description Length (Characters)')
axes[0,0].set_xlabel('Characters')

# Word count distribution
axes[0,1].hist(df['word_count'], bins=50, alpha=0.7)
axes[0,1].set_title('Distribution of Description Length (Words)')
axes[0,1].set_xlabel('Words')

# Length by genre
sns.boxplot(data=df, x='genre', y='word_count', ax=axes[1,0])
axes[1,0].set_title('Word Count by Genre')
axes[1,0].tick_params(axis='x', rotation=45)

# Character length by genre
sns.boxplot(data=df, x='genre', y='description_length', ax=axes[1,1])
axes[1,1].set_title('Character Count by Genre')
axes[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print(f"Average description length: {df['description_length'].mean():.1f} characters")
print(f"Average word count: {df['word_count'].mean():.1f} words")
```

## Genre-Specific Word Clouds

```python
# Create word clouds for top genres
top_genres = df['genre'].value_counts().head(6).index

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for i, genre in enumerate(top_genres):
    genre_text = ' '.join(df[df['genre'] == genre]['description'].astype(str))
    
    wordcloud = WordCloud(width=400, height=300, 
                         background_color='white',
                         max_words=100).generate(genre_text)
    
    axes[i].imshow(wordcloud, interpolation='bilinear')
    axes[i].set_title(f'{genre.title()} Genre', fontsize=14)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```

## Sample Descriptions by Genre

```python
# Show sample descriptions for each genre
for genre in df['genre'].unique()[:5]:  # Show first 5 genres
    print(f"\n{'='*60}")
    print(f"GENRE: {genre.upper()}")
    print(f"{'='*60}")
    
    samples = df[df['genre'] == genre].sample(min(3, len(df[df['genre'] == genre])))
    
    for idx, row in samples.iterrows():
        print(f"\nTitle: {row['title']}")
        print(f"Description: {row['description'][:300]}...")
        print("-" * 40)
```
